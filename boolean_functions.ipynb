```python
# Perceptron
def perceptron(x1, x2, w1, w2, b):
    if (w1*x1 + w2*x2 + b) >= 0 :
        return 1
    else : 
        return 0

# FALSE
def f1(a, b):
    return perceptron(a, b, 0, 0, -1)

# TRUE
def f2(a, b):
    return perceptron(a, b, 0, 0, 1)

def f3(a, b):
    return perceptron(a, b, 1, 1, -1.5)

# OR
def f4(a, b):
    return perceptron(a, b, 1, 1, -0.5)

# NAND
def f5(a, b):
    return perceptron(a, b, -1, -1, 1.5)

# NOR
def f6(a, b):
    return perceptron(a, b, -1, -1, 0.5)

# NOT A
def f7(a, b):
    return perceptron(a, b, -1, 0, 0.5)

# NOT B
def f8(a, b):
    return perceptron(a, b, 0, -1, 0.5)

# A
def f9(a, b):
    return perceptron(a, b, 1, 0, -0.5)

# B
def f10(a, b):
    return perceptron(a, b, 0, 1, -0.5)

# NOT A OR B
def f11(a, b):
    return perceptron(a, b, -1, 1, 0.5)

# A OR NOT B
def f12(a, b):
    return perceptron(a, b, 1, -1, 0.5)

# NOT A AND B
def f13(a, b):
    return perceptron(a, b, -1, 1, -0.5)

# A AND NOT B
def f14(a, b):
    return perceptron(a, b, 1, -1, -0.5)

#XOR
def f15(a, b):
    h1 = perceptron(a, b, 1, -1, -0.5)   # A AND NOT B
    h2 = perceptron(a, b, -1, 1, -0.5)   # NOT A AND B
    return perceptron(h1, h2, 1, 1, -0.5)

# XNOR
def f16(a, b):
    h1 = perceptron(a, b, 1, 1, -1.5)    # AND
    h2 = perceptron(a, b, -1, -1, 0.5)   # NOR
    return perceptron(h1, h2, 1, 1, -0.5)
```

```python
functions = [
    ("FALSE", f1),
    ("TRUE", f2),
    ("AND", f3),
    ("OR", f4),
    ("NAND", f5),
    ("NOR", f6),
    ("NOT A", f7),
    ("NOT B", f8),
    ("A", f9),
    ("B", f10),
    ("NOT A OR B", f11),
    ("A OR NOT B", f12),
    ("NOT A AND B", f13),
    ("A AND NOT B", f14),
    ("XOR", f15),
    ("XNOR", f16)
]

for name, func in functions:
    print("=" * 15)
    print(name)
    print("A B | Output")
    print("-" * 15)

    for a, b in [(0,0), (0,1), (1,0), (1,1)]:
        print(a, b, "|", func(a, b))

    print()
```

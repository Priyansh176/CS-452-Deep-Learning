```python
import numpy as np
```

```python
# Inputs
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])

# Target outputs
y = np.array([0, 0, 0, 1])
```

```python
weights = np.zeros(2)   # w1=0 w2=0
bias = 0                # b = 0

learning_rate = 0.1
epochs = 10

print("Initial Weights:", weights)
print("Initial Bias:", bias)
```

```python
for i in range(epochs):
    print(f"\nEpoch {i + 1}")

    for i in range(len(X)):
        net = np.dot(X[i], weights) + bias  # w1*x1 + w2*x2 + b

        # Activation function
        prediction = 1 if net >= 0 else 0

        error = y[i] - prediction

        weights += learning_rate * error * X[i]    # w_new = w_old +  eta * t * x_i
        bias += learning_rate * error              # b_new = b_old + eta * x_i

        print(f"Input: {X[i]}, Target: {y[i]}, Pred: {prediction}, Error: {error}")

print("\nTraining Complete")
print("Final Weights:", weights)
print("Final Bias:", bias)
```

```python
print("Testing Perceptron\n")

for i in range(len(X)):
    net = np.dot(X[i], weights) + bias
    prediction = 1 if net >= 0 else 0

    print(f"Input: {X[i]}  --> Predicted: {prediction}")
```
